```python
df = sns.load_dataset("mpg")
df
```

```python
df = df.drop("name", axis=1)
df
```

```python
df.origin.unique()
```

```python
# since we have not done the outlier treatment lets replace the value with median
med = df.horsepower.median()
df["horsepower"] = df.horsepower.fillna(med)
```

```python
df.isnull().sum()
```

```python
df.origin.value_counts()
```

```python
new_df = pd.get_dummies(df["origin"])
new_df
```

```python
# new_df = pd.get_dummies(df["origin"])
# df_new = new_df.astype("int64")

# df = pd.concat([df, df_new], axis=1)
# df = df.drop("origin", axis=1)

# df = df.drop("europe", axis=1)


df = pd.get_dummies(
    df,
    columns=["origin"],
    drop_first=True,
    dtype=int
)
```

```python
df
```

```python
# we have done with encoding part 
```

```python
# find what x and y 
X = df.drop("mpg", axis=1)
y = df["mpg"]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
# StandardScaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

regression = LinearRegression()

regression.fit(X_train_scaled, y_train)
```

```python
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": regression.coef_
})

coef_df
```

```python
# observation :
# coeff are very small 
# if one independent variable changes there is not much difference in rediction 
# this is something call smoother model
```

```python
y_pred = regression.predict(X_test_scaled)
y_pred[:5]
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("R2 Score :", r2_score(y_test, y_pred))
print("MAE      :", mean_absolute_error(y_test, y_pred))
print("MSE      :", mean_squared_error(y_test, y_pred))
```

**Ridge Regression**

```python
from sklearn.linear_model import Ridge
ridge = Ridge(alpha=1.0)
# controls regularization strength.

# alpha = 0 → behaves like Linear Regression
# small alpha → weak regularization
# large alpha → strong regularization
```

```python
# train 
ridge.fit(X_train_scaled, y_train)
```

```python
ridge_coeff = pd.DataFrame({
    "Features" : X.columns,
    "Coeff" : ridge.coef_
}
)
ridge_coeff
```

```python
""" 
| Feature      | Linear |  Ridge |
| ------------ | -----: | -----: |
| cylinders    | -0.205 | -0.172 |
| displacement |  1.645 |  1.400 |
| horsepower   | -0.412 | -0.461 |
| weight       | -5.695 | -5.477 |
| acceleration |  0.207 |  0.161 |
| model_year   |  2.913 |  2.884 |
| origin_japan | -0.035 | -0.013 |
| origin_usa   | -1.443 | -1.406 |

"""
```

```python
from sklearn.metrics import r2_score

print("Linear Regression R2 :",
      r2_score(y_test, regression.predict(X_test_scaled)))

print("Ridge R2 :",
      r2_score(y_test, ridge.predict(X_test_scaled)))
```

```python
for alpha in [0.01, 0.1, 1, 10, 100]:
    ridge = Ridge(alpha=alpha)
    ridge.fit(X_train_scaled, y_train)

    print(
        f"Alpha={alpha}",
        "R2=",
        round(r2_score(y_test,
                       ridge.predict(X_test_scaled)),4)
    )
```

```python
""" 
Small alpha
≈ Linear Regression

Large alpha
→ Strong coefficient shrinkage

Very large alpha
→ Underfitting
"""
```

```python
#Laaso regresison
from sklearn.linear_model import Lasso 
lasso = Lasso(alpha=1.0)
```

```python
lasso
```

```python
lasso.fit(X_train_scaled, y_train)
```

```python
y_pred_lasso = lasso.predict(X_test_scaled)
```

```python
lasso_coeff = pd.DataFrame({
    "Feature" : X.columns,
    "Coeff": lasso.coef_
})
lasso_coeff
```

```python
"""
model can predict MPG reasonably well using:

weight
model_year
origin_usa
horsepower
"""
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

print("R2 Score :", r2_score(y_test, y_pred_lasso))
print("MAE      :", mean_absolute_error(y_test, y_pred_lasso))
print("MSE      :", mean_squared_error(y_test, y_pred_lasso))
print("RMSE     :", np.sqrt(mean_squared_error(y_test, y_pred_lasso)))
```

```python
""" 
Linear Regression R2 : 0.8430426014268103
Ridge R2 : 0.842362621546747
Lasso R2: 0.8372767878097774
"""
```

```python
# Elastic net 
from sklearn.linear_model import ElasticNet
elastic = ElasticNet(
    alpha=1.0,
    l1_ratio=0.5
)
"""
l1_ratio = 0
    ↓
Pure Ridge

l1_ratio = 1
    ↓
Pure Lasso

l1_ratio = 0.5
    ↓
50% Ridge + 50% Lasso
"""
```

```python
elastic.fit(X_train_scaled, y_train)
```

```python
elastic_coeff = pd.DataFrame({
    "Feature": X.columns,
    "Coeff": elastic.coef_
})

elastic_coeff
```

```python
y_pred_elastic = elastic.predict(X_test_scaled)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

print("R2 Score :", r2_score(y_test, y_pred_elastic))
print("MAE      :", mean_absolute_error(y_test, y_pred_elastic))
print("MSE      :", mean_squared_error(y_test, y_pred_elastic))
print("RMSE     :", np.sqrt(mean_squared_error(y_test, y_pred_elastic)))
```

```python
""" 
Linear Regression R2 : 0.8430426014268103
Ridge R2 : 0.842362621546747
Lasso R2: 0.8372767878097774
ElasticNet R2 : 0.8093823034042484
"""
```

```python
"""
First Observation

For this dataset:

Linear Regression ≈ Ridge > Lasso > Elastic Net

This tells us something important.

There is no major overfitting problem

If Ridge were significantly better than Linear Regression:

Linear = 0.78
Ridge  = 0.84

then regularization would be helping a lot.

But here:

0.8430
vs
0.8424

Difference is tiny.

So Linear Regression is already doing quite well.
"""
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
lasso = Lasso()
param_grid = {
    "alpha" : [0.001, 0.01, 0.1, 1, 10, 100]
}
grid_search = GridSearchCV(
    estimator= lasso,
    param_grid= param_grid,
    scoring= "r2",
    verbose=2,
    cv=5
)
grid_search
```

```python
grid_search.fit(X_train_scaled, y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
grid_search.cv_results_
```

```python
# best_estimator_ : estimator that was chosen by the search, i.e. estimator which gave highest score (or 
#  small loss if specified) on the left out data. Not available if refit = False
```

```python
# grid_search.best_estimator_
best_lasso = grid_search.best_estimator_
y_pred = best_lasso.predict(X_test_scaled)
```

```python
lasso_r2 = r2_score(y_test, y_pred)
lasso_r2
```

```python
# randomized search : only param_grid change to param_distributon 
```

```python
from sklearn.model_selection import RandomizedSearchCV

lasso = Lasso()

param_distributions = {
    'alpha': [0.001, 0.01, 0.1, 1, 10, 100]
}

random_search = RandomizedSearchCV(
    estimator=lasso,
    param_distributions=param_distributions,
    n_iter=5,
    scoring='r2',
    cv=5,
    verbose=2,
    random_state=42
)
random_search
```

```python
""" 
n_iter : select random n values from param_distributions 
random_state : reproduce the result 
"""
```

```python
random_search.fit(X_train_scaled, y_train)
```

```python
random_search.best_params_
```

```python
random_search.best_score_
```

```python
best_lasso_randomized = random_search.best_estimator_
```
