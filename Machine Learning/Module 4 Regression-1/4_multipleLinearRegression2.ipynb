```python
from sklearn.datasets import fetch_california_housing

# Load the California housing dataset
# Doubt: why use fetch_california_housing?
# Answer: it is a built-in regression dataset from sklearn, suitable for multiple linear regression.
data = fetch_california_housing()

data
```

```python
# The dataset is stored in key-value style
# Doubt: what does keys() do?
# Answer: it shows what information is available inside the dataset object.
data.keys()
```

```python
# Feature names tell us the input columns
data.feature_names
```

```python
# Raw feature matrix
data.data
```

```python
# Raw feature matrix
data.data
```

```python
import pandas as pd

# Convert feature matrix into a DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

df.head()
```

```python
# Add the target column to the DataFrame
# I am naming it 'HousePrice' for clarity.
# Doubt: why add target as a column?
# Answer: it becomes easier to split X and Y and understand the dataset visually.
df["HousePrice"] = data.target

df.head()
```

```python
# Dataset shape
df.shape
```

```python
# Dataset shape
df.shape
```

```python
# Statistical summary
df.describe()
```

```python
# X = all independent variables except target
# Y = target variable
# Doubt: why remove target from X?
# Answer: because the model should learn from features only, not from the answer column.
X = df.drop("HousePrice", axis=1)
Y = df["HousePrice"]

print(X.head())
print(Y.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
# Doubt: why do we split?
# Answer: training data teaches the model, testing data checks whether the model works on unseen data.
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape :", Y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler object
scaler = StandardScaler()

# Fit on training data only and transform both train/test
# Doubt: why fit only on training data?
# Answer: to avoid data leakage. The test set must remain unseen.
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train_scaled[:5])
```

```python
from sklearn.linear_model import LinearRegression

# Create the regression model
regressor = LinearRegression()

regressor
```

```python
# Train the model on scaled training data
# Doubt: why use X_train_scaled and Y_train?
# Answer: because training should happen only on training data, and scaling makes feature magnitudes consistent.
regressor.fit(X_train_scaled, Y_train)
```

```python
# Intercept
regressor.intercept_
```

```python
# Feature-wise coefficient view
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": regressor.coef_
})

coef_df
```

```python
# Predict on test data
Y_test_pred = regressor.predict(X_test_scaled)

Y_test_pred[:10]
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(Y_test, Y_test_pred)
mse = mean_squared_error(Y_test, Y_test_pred)
rmse = mse ** 0.5
r2 = r2_score(Y_test, Y_test_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
# Adjusted R²
# Doubt: why use adjusted R² here?
# Answer: because this is multiple linear regression, and adjusted R² penalizes useless extra features.
n = X_test_scaled.shape[0]   # number of rows
p = X_test_scaled.shape[1]   # number of features

adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)

print("Adjusted R²:", adjusted_r2)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Residuals = actual - predicted
# Doubt: why do we check residuals?
# Answer: to see whether errors are centered around zero and whether the model is behaving properly.
residuals = Y_test - Y_test_pred

sns.histplot(residuals, kde=True)
plt.title("Residual Distribution")
plt.xlabel("Residuals")
plt.show()
```

```python
# Residual scatter plot
plt.scatter(Y_test_pred, residuals)
plt.axhline(y=0, color="red")

plt.xlabel("Predicted House Price")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.show()
```
